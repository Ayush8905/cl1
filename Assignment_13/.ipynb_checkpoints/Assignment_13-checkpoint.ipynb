# Assignment 13 – Time Series Analysis: Stock Price Forecasting

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.seasonal import seasonal_decompose
```

## 1. Import & Explore

```python
df = pd.read_csv('Stock_Prices.csv')
print(df.shape)
df.head(3)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').set_index('Date')
print(df.dtypes)
print('\nDate range:', df.index.min(), '→', df.index.max())
```

## 2. Historical Price Trend

```python
plt.figure(figsize=(12, 4))
plt.plot(df.index, df['AAPL.Close'], color='steelblue', label='Close')
plt.fill_between(df.index, df['AAPL.Low'], df['AAPL.High'], alpha=0.2, color='steelblue')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.title('AAPL Historical Close Price')
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Moving Averages

```python
df['MA_20']  = df['AAPL.Close'].rolling(window=20).mean()
df['MA_50']  = df['AAPL.Close'].rolling(window=50).mean()

plt.figure(figsize=(12, 4))
plt.plot(df.index, df['AAPL.Close'], label='Close',  color='steelblue', alpha=0.6)
plt.plot(df.index, df['MA_20'],      label='20-day MA', color='tomato')
plt.plot(df.index, df['MA_50'],      label='50-day MA', color='seagreen')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.title('AAPL Close – Moving Averages')
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. Seasonality Analysis

```python
decomp = seasonal_decompose(df['AAPL.Close'].dropna(), model='additive', period=20)

fig, axes = plt.subplots(4, 1, figsize=(12, 10))
decomp.observed.plot(ax=axes[0], color='steelblue');  axes[0].set_title('Observed')
decomp.trend.plot(ax=axes[1],    color='tomato');      axes[1].set_title('Trend')
decomp.seasonal.plot(ax=axes[2], color='seagreen');    axes[2].set_title('Seasonal')
decomp.resid.plot(ax=axes[3],    color='gray');        axes[3].set_title('Residual')

plt.suptitle('Seasonal Decomposition – AAPL Close')
plt.tight_layout()
plt.show()
```

## 5. Correlation – Close vs Volume

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].scatter(df['AAPL.Volume'], df['AAPL.Close'], alpha=0.4, s=10, color='orchid')
axes[0].set_xlabel('Volume')
axes[0].set_ylabel('Close Price ($)')
axes[0].set_title('Close vs Volume')

price_cols = ['AAPL.Open','AAPL.High','AAPL.Low','AAPL.Close','AAPL.Adjusted']
sns.heatmap(df[price_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=axes[1])
axes[1].set_title('Price Columns Correlation')

plt.suptitle('AAPL – Correlation Analysis')
plt.tight_layout()
plt.show()
```

## 6. ARIMA Forecast

```python
close = df['AAPL.Close'].dropna()
train = close[:-30]
test  = close[-30:]

model = ARIMA(train, order=(5, 1, 0))
fitted = model.fit()
forecast = fitted.forecast(steps=30)

plt.figure(figsize=(12, 4))
plt.plot(train.index, train,        label='Train',    color='steelblue')
plt.plot(test.index,  test,         label='Actual',   color='seagreen')
plt.plot(test.index,  forecast,     label='Forecast', color='tomato', linestyle='--')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.title('ARIMA(5,1,0) – 30-Day Forecast')
plt.legend()
plt.tight_layout()
plt.show()

rmse = np.sqrt(((forecast.values - test.values)**2).mean())
print(f'RMSE: {rmse:.4f}')
```
