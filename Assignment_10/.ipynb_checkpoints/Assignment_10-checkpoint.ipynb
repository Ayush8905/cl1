# Assignment 10 – Data Visualisation: Air Quality Index (AQI) Trends

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Import & Explore

```python
df = pd.read_csv('City_Air_Quality.csv')
print(df.shape)
df.head(3)
```

```python
print('Missing values:')
print(df.isnull().sum()[df.isnull().sum() > 0])

df['Date'] = pd.to_datetime(df['Date'])
df = df.dropna(subset=['AQI'])
df['Year']  = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Fill pollutant NaNs with column median
pollutants = ['PM2.5', 'PM10', 'NO', 'NO2', 'CO', 'SO2', 'O3']
for col in pollutants:
    df[col] = df[col].fillna(df[col].median())

print('Shape after cleaning:', df.shape)
```

## 2. Overall AQI Trend Over Time

```python
# Pick one city for trend clarity
city = df['City'].value_counts().index[0]
df_city = df[df['City'] == city].sort_values('Date')

monthly_aqi = df_city.groupby(['Year', 'Month'])['AQI'].mean().reset_index()
monthly_aqi['period'] = pd.to_datetime(monthly_aqi[['Year','Month']].assign(day=1))

plt.figure(figsize=(12, 4))
plt.plot(monthly_aqi['period'], monthly_aqi['AQI'], color='tomato')
plt.xlabel('Date')
plt.ylabel('AQI')
plt.title(f'Monthly AQI Trend – {city}')
plt.tight_layout()
plt.show()
```

## 3. Individual Pollutant Trends

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
colours = ['steelblue','seagreen','tomato','orchid','gold','coral']

for ax, pol, col in zip(axes.flat, ['PM2.5','PM10','CO','SO2','NO2','O3'], colours):
    monthly = df_city.groupby(['Year','Month'])[pol].mean().reset_index()
    monthly['period'] = pd.to_datetime(monthly[['Year','Month']].assign(day=1))
    ax.plot(monthly['period'], monthly[pol], color=col)
    ax.set_title(pol)
    ax.set_xlabel('Date')
    ax.set_ylabel('Level')
    ax.tick_params(axis='x', rotation=30)

plt.suptitle(f'Pollutant Trends Over Time – {city}')
plt.tight_layout()
plt.show()
```

## 4. AQI Across AQI Buckets – Bar Plot

```python
bucket_counts = df['AQI_Bucket'].value_counts()

plt.figure(figsize=(8, 4))
plt.bar(bucket_counts.index, bucket_counts.values, color='cornflowerblue', edgecolor='k', linewidth=0.4)
plt.xlabel('AQI Bucket')
plt.ylabel('Count')
plt.title('AQI Bucket Distribution (All Cities)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 5. AQI Distribution – Box Plot by AQI Bucket

```python
bucket_order = df.groupby('AQI_Bucket')['AQI'].median().sort_values().index

plt.figure(figsize=(10, 5))
df.boxplot(column='AQI', by='AQI_Bucket',
           positions=range(len(bucket_order)),
           labels=bucket_order)
plt.xlabel('AQI Bucket')
plt.ylabel('AQI')
plt.title('AQI Distribution by Bucket')
plt.suptitle('')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 6. AQI vs Pollutants – Scatter Plot

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, pol, col in zip(axes, ['PM2.5','PM10','NO2'], ['steelblue','seagreen','tomato']):
    ax.scatter(df[pol], df['AQI'], alpha=0.2, s=5, c=col)
    ax.set_xlabel(pol)
    ax.set_ylabel('AQI')
    ax.set_title(f'AQI vs {pol}')

plt.suptitle('AQI vs Pollutant Levels')
plt.tight_layout()
plt.show()
```

## 7. Correlation Heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[pollutants + ['AQI']].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Pollutant vs AQI Correlation')
plt.tight_layout()
plt.show()
```
