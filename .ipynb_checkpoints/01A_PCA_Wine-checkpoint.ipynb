# Assignment 1A - PCA on Wine

The PDF links a wine cultivar dataset (three classes), although its text asks about red/white wine. Both are covered below. The second dataset is UCI Wine Quality, which has actual red/white labels. PCA keeps at least 90% of the variance; the scatter plots show only its first two components.

```python
# Load the wine file linked in the assignment.
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

wine = pd.read_csv('datasets/assignment_01/part_a/wine_link.csv')
display(wine.head())
```

```python
# Standardize measurements and keep components explaining 90% of the variance.
X = wine.drop(columns='Customer_Segment')
X_scaled = StandardScaler().fit_transform(X)
pca = PCA(n_components=0.90)
components = pca.fit_transform(X_scaled)
print('Original features:', X.shape[1])
print('Principal components:', pca.n_components_)
print('Variance retained:', pca.explained_variance_ratio_.sum())
```

```python
# Plot the first two components, colored by the actual cultivar label.
for label in sorted(wine['Customer_Segment'].unique()):
    rows = wine['Customer_Segment'] == label
    plt.scatter(components[rows, 0], components[rows, 1], label=f'Cultivar {label}')
plt.xlabel('Principal component 1')
plt.ylabel('Principal component 2')
plt.title('PCA: linked wine dataset')
plt.legend()
plt.show()
```

## Red and white wine
The quality score is excluded from the chemical measurements. Color labels are used only to color the plot.

```python
# Combine the downloaded red and white wine files.
red = pd.read_csv('datasets/assignment_01/part_a/winequality-red.csv', sep=';')
white = pd.read_csv('datasets/assignment_01/part_a/winequality-white.csv', sep=';')
red['type'] = 'red'
white['type'] = 'white'
wines = pd.concat([red, white], ignore_index=True)
X = wines.drop(columns=['type', 'quality'])
X_scaled = StandardScaler().fit_transform(X)
pca = PCA(n_components=0.90)
components = pca.fit_transform(X_scaled)
print('Features before PCA:', X.shape[1])
print('Components after PCA:', pca.n_components_)
print('Variance retained:', pca.explained_variance_ratio_.sum())
```

```python
# Compare red and white wine in the first two principal components.
for label, color in [('red', 'firebrick'), ('white', 'goldenrod')]:
    rows = wines['type'] == label
    plt.scatter(components[rows, 0], components[rows, 1], s=10, alpha=0.4,
                label=label, color=color)
plt.xlabel('Principal component 1')
plt.ylabel('Principal component 2')
plt.title('PCA: red and white wine')
plt.legend()
plt.show()
```
