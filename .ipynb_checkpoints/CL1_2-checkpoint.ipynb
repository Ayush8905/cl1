```python
import pandas as pd
```

```python
csv_data = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard"],
    "Sales": [50000, 1000, 2000]
})
csv_data.to_csv("sales.csv", index=False)

excel_data = pd.DataFrame({
    "Product": ["Laptop", "Monitor", "Mouse"],
    "Sales": [45000, 15000, 1200]
})
excel_data.to_excel("sales.xlsx", index=False)

json_data = pd.DataFrame({
    "Product": ["Keyboard", "Monitor", "Tablet"],
    "Sales": [2500, 18000, 22000]
})
json_data.to_json("sales.json", orient="records")
```

```python
csv_df = pd.read_csv("sales.csv")
excel_df = pd.read_excel("sales.xlsx")
json_df = pd.read_json("sales.json")

print("CSV Data:")
print(csv_df)

print("\nExcel Data:")
print(excel_df)

print("\nJSON Data:")
print(json_df)
```

```python
for df in [csv_df, excel_df, json_df]:
    df["Product"] = df["Product"].str.strip().str.title()
    df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
    df.drop_duplicates(inplace=True)
    df.dropna(inplace=True)

print("Data cleaned successfully.")
```

```python
sales = pd.concat(
    [csv_df, excel_df, json_df],
    ignore_index=True
)

print(sales)
```

```python
print("Total Sales:", sales["Sales"].sum())
print("\nAverage Sales:", sales["Sales"].mean())

print("\nSales by Product:")
print(sales.groupby("Product")["Sales"].sum())
```

```python
import matplotlib.pyplot as plt

sales.groupby("Product")["Sales"].sum().plot(
    kind="bar",
    title="Sales by Product"
)

plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales.to_csv("unified_sales.csv", index=False)
```
