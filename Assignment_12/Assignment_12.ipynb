# Assignment 12 – Data Aggregation: Retail Sales Performance by Region

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Import & Explore

```python
df = pd.read_csv('Retail_Sales_Data.csv')
print(df.shape)
df.head(3)
```

```python
print('Missing values:')
print(df.isnull().sum()[df.isnull().sum() > 0])

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year
df = df.dropna(subset=['Sales', 'Region', 'Category'])
print('Shape after cleaning:', df.shape)
```

## 2. Identify Relevant Variables

```python
print('Regions  :', df['Region'].unique())
print('Categories:', df['Category'].unique())
print('Years    :', sorted(df['Year'].unique()))
```

## 3. Total Sales by Region

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(region_sales)
print(f'\nTop-performing region: {region_sales.idxmax()}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(region_sales.index, region_sales.values, color='steelblue', edgecolor='k', linewidth=0.4)
axes[0].set_title('Total Sales by Region')
axes[0].set_ylabel('Sales ($)')

axes[1].pie(region_sales.values, labels=region_sales.index, autopct='%1.1f%%')
axes[1].set_title('Sales Distribution by Region')

plt.suptitle('Regional Sales Overview')
plt.tight_layout()
plt.show()
```

## 4. Sales by Region × Product Category

```python
region_cat = df.groupby(['Region', 'Category'])['Sales'].sum().unstack()
print(region_cat)
```

```python
region_cat.plot(kind='bar', figsize=(9, 5), edgecolor='k', linewidth=0.4)
plt.title('Sales by Region and Product Category')
plt.ylabel('Sales ($)')
plt.xlabel('Region')
plt.xticks(rotation=0)
plt.legend(title='Category')
plt.suptitle('')
plt.tight_layout()
plt.show()
```

## 5. Top-Performing Regions & Stacked Bar

```python
region_cat.plot(kind='bar', stacked=True, figsize=(9, 5),
                color=['steelblue','seagreen','tomato'], edgecolor='k', linewidth=0.4)
plt.title('Stacked Sales by Region and Category')
plt.ylabel('Sales ($)')
plt.xlabel('Region')
plt.xticks(rotation=0)
plt.legend(title='Category')
plt.suptitle('Regional × Category Sales Breakdown')
plt.tight_layout()
plt.show()
```

```python
# Additional metrics
print('Sales summary by Region:')
print(df.groupby('Region')['Sales'].agg(['sum','mean','count']).round(2))

print('\nTop 5 Sub-Categories by Sales:')
print(df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head())
```
