# Assignment 6 – Data Cleaning and Preparation: Telecom Customer Churn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

## 1. Import & Explore

```python
df = pd.read_csv('Telecom_Customer_Churn.csv')
print(df.shape)
df.head(3)
```

```python
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum()[df.isnull().sum() > 0])
```

## 2. Handle Missing Values & Fix Data Types

```python
# TotalCharges is stored as string — convert, coerce blanks to NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print('NaNs in TotalCharges after conversion:', df['TotalCharges'].isna().sum())

# Fill with median (tenure=0 customers have no charges)
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
```

## 3. Remove Duplicates

```python
print('Duplicates before:', df.duplicated().sum())
df = df.drop_duplicates()
print('Duplicates after :', df.duplicated().sum())
```

## 4. Check & Fix Inconsistent Data

```python
# Standardise string columns — strip whitespace, title-case
str_cols = df.select_dtypes('object').columns.tolist()
for col in str_cols:
    df[col] = df[col].str.strip()

# Check unique values for binary-ish columns
for col in ['gender', 'Partner', 'Dependents', 'Churn', 'PhoneService', 'PaperlessBilling']:
    print(col, ':', df[col].unique())
```

## 5. Identify & Handle Outliers

```python
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, col in zip(axes, num_cols):
    ax.boxplot(df[col].dropna())
    ax.set_title(col)
plt.suptitle('Outlier Check – Numeric Features')
plt.tight_layout()
plt.show()
```

## 6. Feature Engineering

```python
# Bin tenure into customer segments
df['tenure_group'] = pd.cut(df['tenure'],
                            bins=[0, 12, 24, 48, 72],
                            labels=['0-1yr', '1-2yr', '2-4yr', '4-6yr'])

# Flag high-value customers
df['high_value'] = (df['MonthlyCharges'] > df['MonthlyCharges'].quantile(0.75)).astype(int)

df[['tenure', 'tenure_group', 'MonthlyCharges', 'high_value']].head(4)
```

## 7. Encode Categorical Variables

```python
df_encoded = df.drop(columns=['customerID', 'tenure_group'])

le = LabelEncoder()
for col in df_encoded.select_dtypes('object').columns:
    df_encoded[col] = le.fit_transform(df_encoded[col])

df_encoded.head(3)
```

## 8. Normalise & Split

```python
X = df_encoded.drop('Churn', axis=1)
y = df_encoded['Churn']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
print('Train:', X_train.shape, '| Test:', X_test.shape)
```

## 9. Export Cleaned Dataset

```python
df_encoded.to_csv('Telecom_Cleaned.csv', index=False)
print('Saved: Telecom_Cleaned.csv')

# Churn distribution
churn_counts = df['Churn'].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].bar(churn_counts.index, churn_counts.values, color=['steelblue', 'tomato'])
axes[0].set_title('Churn Distribution')
axes[0].set_ylabel('Count')

axes[1].pie(churn_counts.values, labels=churn_counts.index, autopct='%1.1f%%')
axes[1].set_title('Churn Proportion')

plt.suptitle('Telecom Customer Churn Overview')
plt.tight_layout()
plt.show()
```
