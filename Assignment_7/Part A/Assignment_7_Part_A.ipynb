# Assignment 7A – K-Means Clustering on Iris Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv('Iris_1.csv')
print(df.shape)
df.head(3)
```

```python
feature_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
X = df[feature_cols].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Elbow Method – Determine Optimal k

```python
inertias = []
k_range  = range(1, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertias.append(km.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(k_range, inertias, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method – Iris')
plt.tight_layout()
plt.show()
```

## Fit K-Means (k=3)

```python
km = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = km.fit_predict(X_scaled)

sil = silhouette_score(X_scaled, df['Cluster'])
print(f'Silhouette Score: {sil:.4f}')
```

```python
# Cluster vs true species comparison
print(pd.crosstab(df['Cluster'], df['Species']))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
colours = ['tomato', 'steelblue', 'seagreen']

# Clusters
for c, col in zip([0, 1, 2], colours):
    mask = df['Cluster'] == c
    axes[0].scatter(X_scaled[mask, 0], X_scaled[mask, 2],
                    c=col, label=f'Cluster {c}', edgecolors='k', linewidths=0.3)
axes[0].scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 2],
                marker='X', s=150, c='gold', edgecolors='k', linewidths=0.5, label='Centroids')
axes[0].set_title('K-Means Clusters')
axes[0].set_xlabel('SepalLength (scaled)')
axes[0].set_ylabel('PetalLength (scaled)')
axes[0].legend()

# True labels
species_list = df['Species'].unique()
for sp, col in zip(species_list, colours):
    mask = df['Species'] == sp
    axes[1].scatter(X_scaled[mask, 0], X_scaled[mask, 2],
                    c=col, label=sp, edgecolors='k', linewidths=0.3)
axes[1].set_title('True Species')
axes[1].set_xlabel('SepalLength (scaled)')
axes[1].set_ylabel('PetalLength (scaled)')
axes[1].legend()

plt.suptitle('K-Means Clustering vs True Labels – Iris')
plt.tight_layout()
plt.show()
```
